## NaN Handling Strategy for Meteostat Weather Variables

We need to handle missing values (`NaN`) in **data_delhi** and **data_kolkata** using
**domain-aware imputation strategies**, based on the physical meaning of each variable.

---

### Weather Variables Description

- **dwpt** → Dew point temperature (°C)  
- **temp** → Air temperature (°C)  
- **rhum** → Relative humidity (%)  
- **prcp** → Precipitation (mm)  
- **wdir** → Wind direction (degrees)  
- **wspd** → Wind speed (km/h)  
- **pres** → Air pressure (hPa)  
- **coco** → Weather condition code  
  - e.g., Fair = 1, Thunderstorm = 12  

---

### NaN Handling Strategy

| Column | Meaning | Data Type | What I decided to use | Reasoning |
|------|--------|----------|----------------------|-----------|
| **temp** | Air temperature (°C) | Continuous | Time interpolation (`interpolate(method="time")`) | Temperature changes smoothly over time; interpolation preserves trends |
| **dwpt** | Dew point (°C) | Continuous | Time interpolation | Dew point is strongly correlated with temperature and humidity |
| **rhum** | Relative humidity (%) | Continuous (bounded 0–100) | Forward fill (short gaps) + clipping | Humidity is bounded and typically stable over short intervals |
| **prcp** | Precipitation (mm) | Zero-inflated | Fill NaN with `0` | Missing precipitation usually implies no rainfall; ML or averaging is incorrect |
| **wdir** | Wind direction (°) | Circular |Forward fill | The wind direction will be sort of same to of the previous day|
| **wspd** | Wind speed (km/h) | Continuous | Time interpolation | Wind speed varies smoothly but may spike; interpolation works for short gaps |
| **pres** | Air pressure (hPa) | Continuous | Linear interpolation | Pressure changes slowly and predictably over time |
| **coco** | Weather condition code | Categorical | Forward fill | Weather condition labels persist for short durations and are categorical |

---

### Key Notes

- RandomForest-based imputation was considered **only for continuous variables**
  such as `temp`, `dwpt`, `wspd`, and `rhum`.
- ML-based imputation was **not used** for `prcp`, `wdir`, or `coco`
  due to their physical and statistical properties.
- Domain knowledge was prioritized over blind statistical filling.

---

### Summary

> NaN handling was performed using domain-aware decisions rather than a single
imputation method, ensuring that each weather variable was treated according to
its physical meaning and statistical behavior.


## calculating nan values for kolkata and delhi

```python
data_kolkata.isna().sum()
```

```python
data_delhi.isna().sum()
```

```python
data_delhi["wdir"] = data_delhi["wdir"].ffill(limit=2).bfill(limit=2)
data_kolkata["wdir"] = data_kolkata["wdir"].ffill(limit=2).bfill(limit=2)
```

```python
data_delhi["coco"] = data_delhi["coco"].ffill()
data_kolkata["coco"] = data_kolkata["coco"].ffill()
```

```python
data_delhi["prcp"] = data_delhi["prcp"].fillna(0)
data_kolkata["prcp"] = data_kolkata["prcp"].fillna(0)
```
